```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
```

```python
# Load the cleaned dataset
df = pd.read_csv("cleaned_monkeypox_tweets.csv")
```

```python
# Display the first 5 rows of the dataset
print("Cleaned Dataset:")
print(df.head())
```

```python
# Step 1: Lowercasing
df["Post description"] = df["Post description"].str.lower()
df["Translated_Post_Description"] = df["Translated_Post_Description"].str.lower()
```

```python
# Modified function to handle NaN and non-string values
def remove_special_chars(text):
    if pd.isna(text):  # Check for NaN values
        return ""  # Return empty string for NaN
    return re.sub(r"[^a-zA-Z0-9\s]", "", str(text))  # Convert to string and clean

# Apply function to both columns
df["Post description"] = df["Post description"].apply(remove_special_chars)
df["Translated_Post_Description"] = df["Translated_Post_Description"].apply(remove_special_chars)

print(df)
```

```python
# Step 3: Tokenization
df["Post description_tokens"] = df["Post description"].apply(word_tokenize)
df["Translated_Post_Description_tokens"] = df["Translated_Post_Description"].apply(word_tokenize)
```

```python
# Step 4: Remove Stopwords
stop_words = set(stopwords.words("english"))

def remove_stopwords(tokens):
    return [word for word in tokens if word not in stop_words]

df["Post description_tokens"] = df["Post description_tokens"].apply(remove_stopwords)
df["Translated_Post_Description_tokens"] = df["Translated_Post_Description_tokens"].apply(remove_stopwords)
print(df)
```

```python
# Step 5: Stemming and Lemmatization
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

def stem_tokens(tokens):
    return [stemmer.stem(word) for word in tokens]

def lemmatize_tokens(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]
```

```python
# Apply stemming
df["Post description_stemmed"] = df["Post description_tokens"].apply(stem_tokens)
df["Translated_Post_Description_stemmed"] = df["Translated_Post_Description_tokens"].apply(stem_tokens)
print(df)
```

```python
# Apply lemmatization
df["Post description_lemmatized"] = df["Post description_tokens"].apply(lemmatize_tokens)
df["Translated_Post_Description_lemmatized"] = df["Translated_Post_Description_tokens"].apply(lemmatize_tokens)
print(df)
```

```python
# Step 6: Vectorization (TF-IDF)
# Combine tokenized text into strings for vectorization
df["Post description_processed"] = df["Post description_lemmatized"].apply(" ".join)
df["Translated_Post_Description_processed"] = df["Translated_Post_Description_lemmatized"].apply(" ".join)
print(df)
```

```python
# Initialize TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer(max_features=150)  # Limit to 1000 features for simplicity

# Fit and transform the text data
tfidf_matrix = tfidf_vectorizer.fit_transform(df["Post description_processed"])
```

```python
# Convert TF-IDF matrix to a DataFrame
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out())
```

```python
# Display the TF-IDF DataFrame
print("\nTF-IDF Matrix:")
print(tfidf_df.head())
```
