```python
import pandas as pd
import numpy as np
import re

```

```python
# Load the dataset
file_path = "C:\\Users\\samir\\Desktop\\M.Tech\\NLP\\Twitter\\Dataset.xlsx"
df = pd.read_excel(file_path)

```

```python
# Display the first 5 rows of the dataset
print("Initial Dataset:")
print(df.head())

```

```python
# Step 1: Explore the Dataset
print("\nDataset Info:")
print(df.info())

```

```python
print("\nMissing Values:")
print(df.isnull().sum())

```

```python
print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Step 2: Handle Missing Values
# Drop rows with missing values in critical columns (e.g., Post description, Date)
df.dropna(subset=["Post description", "Date"], inplace=True)
```

```python
# Step 3: Clean Text Data
def clean_text(text):
    if isinstance(text, str):
        # Remove URLs
        text = re.sub(r"http\S+|www\S+|https\S+", "", text, flags=re.MULTILINE)
        # Remove special characters and numbers
        text = re.sub(r"\W", " ", text)
        text = re.sub(r"\d", " ", text)
        # Convert to lowercase
        text = text.lower()
        # Remove extra spaces
        text = re.sub(r"\s+", " ", text).strip()
    return text

```

```python
# Apply cleaning to text columns
df["Post description"] = df["Post description"].apply(clean_text)
df["Translated Post Description"] = df["Translated Post Description"].apply(clean_text)
```

```python
#Step 4: Convert Data Types
# Convert 'Date' column to datetime format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

```

```python
# Drop rows with invalid dates (if any)
df.dropna(subset=["Date"], inplace=True)

```

```python
# Step 5: Handle Categorical Columns
# Encode 'Language' column (one-hot encoding)
df = pd.get_dummies(df, columns=["Language"], prefix="Lang")

```

```python
# Encode 'Sentiment', 'Hate', and 'Stress or Anxiety' columns (if they are categorical)
# For example, convert 'Sentiment' to numerical values
sentiment_mapping = {"Positive": 1, "Neutral": 0, "Negative": -1}
df["Sentiment"] = df["Sentiment"].map(sentiment_mapping)

```

```python
# Example mapping with a default value for unknown categories
hate_mapping = {"Hate": 1, "Not Hate": 0, "Unknown": 0}
df["Hate"] = df["Hate"].map(hate_mapping).fillna(0).astype(int)

stress_mapping = {"Stress/Anxiety Detected": 1, "No Stress/Anxiety Detected": 0, "Unknown": 0}
df["Stress or Anxiety"] = df["Stress or Anxiety"].map(stress_mapping).fillna(0).astype(int)
```

```python
# Convert 'Hate' and 'Stress or Anxiety' to binary (1 for True, 0 for False)
df["Hate"] = df["Hate"].astype(int)
df["Stress or Anxiety"] = df["Stress or Anxiety"].astype(int)
```

```python
# Step 6: Save the Cleaned Dataset
output_file = "cleaned_monkeypox_tweets.csv"
df.to_csv(output_file, index=False)
```
